# Figure 4.8: Fourier shape descriptors

*The Discrete Algebra of the Fourier Transform*, Gabriel Peyré.

A polygon and a perturbed, rotated version, with their magnitude-based Fourier descriptors. The perturbation changes the descriptor; the rotation alone would leave it unchanged.

[Read this figure in the book](https://mathematical-tours.github.io/daft-sources/DAFT-EN.pdf#page=90).

## Prepare the example

Run the cells in order in Jupyter or Google Colab. The first cell installs missing Python packages and downloads the figure sources and input data once. Subsequent runs use the cached download. Python 3.10 or later is required.

```python
from pathlib import Path
import hashlib
import importlib.util
import io
import os
import subprocess
import sys
import urllib.request
import zipfile

packages = {"numpy": "numpy", "scipy": "scipy", "matplotlib": "matplotlib",
            "PIL": "Pillow", "sympy": "sympy"}
missing = [package for module, package in packages.items()
           if importlib.util.find_spec(module) is None]
if missing:
    subprocess.check_call([sys.executable, "-m", "pip", "install", *missing])

archive_url = os.environ.get("DAFT_FIGURE_ARCHIVE", 'https://mathematical-tours.github.io/daft-sources/downloads/daft-figures-2e843bbe89f0.zip')
archive_sha = '2e843bbe89f0e274849b6896eb16bd9f7686521a8ca98f633240bfc06ea2d5b1'
workspace = Path(os.environ.get("DAFT_FIGURE_WORKDIR", "daft-figure-files")) / archive_sha[:12]
cache = workspace / "sources.zip"
workspace.mkdir(parents=True, exist_ok=True)
data = cache.read_bytes() if cache.exists() else urllib.request.urlopen(archive_url, timeout=60).read()
if hashlib.sha256(data).hexdigest() != archive_sha:
    raise ValueError("The figure-source download is incomplete or has changed. Download it again.")
cache.write_bytes(data)
with zipfile.ZipFile(io.BytesIO(data)) as archive:
    archive.extractall(workspace)
ROOT = (workspace / "daft-figures").resolve()
__file__ = str(ROOT / 'figures-code/chapter-fourier-applications/fourier-descriptors/generate.py')
print("Figure sources ready.")
```

## Generate the figure

This is the plotting code used by the book. Change its parameters and rerun this cell, then the display cell below, to explore the example.

```python
from pathlib import Path
import sys
ROOT=Path(__file__).resolve().parents[3]
sys.path.insert(0,str(ROOT/"tools"))
sys.path.insert(0,str(ROOT/"listings"))
from figure_style import *
from applications_data import polygon,plot_polygon
z=polygon();rng=np.random.default_rng(20260908)
w=z+.025*(rng.standard_normal(len(z))+1j*rng.standard_normal(len(z)))
w=w.mean()+np.exp(.2j*np.pi)*(w-w.mean())
descriptor=lambda x:abs(np.fft.fft(x)[2:])/abs(np.fft.fft(x)[1])
assert np.allclose(descriptor(z),descriptor((2+3j)*z+4))
fig,axes=plt.subplots(1,2,figsize=(8.7,3.8))
plot_polygon(axes[0],z,color=TEAL,label='original');plot_polygon(axes[0],w,color=RED,ls='--',label='perturbed and rotated')
axes[0].set(aspect='equal',title='Polygons');axes[0].legend(frameon=False,fontsize=8);axes_box(axes[0],'real coordinate','imaginary coordinate')
for x,color,marker,label in [(z,TEAL,'o','original'),(w,RED,'x','modified')]:axes[1].plot(np.arange(2,len(z)),descriptor(x),marker,color=color,ms=4,label=label)
axes_box(axes[1],'frequency index $k$',r'$|\widehat\Pi[k]|/|\widehat\Pi[1]|$');axes[1].set_title('Magnitude descriptors');axes[1].legend(frameon=False,fontsize=8)
save(fig,'chapter-fourier-applications','fourier-descriptors')
```
