# Figure 4.14: Template matching by correlation

*The Discrete Algebra of the Fourier Transform*, Gabriel Peyré.

Template matching in the flower image. The red box identifies the extracted patch; crosses mark its location in the raw and normalized correlation maps. Only complete-template positions are included.

[Read this figure in the book](https://mathematical-tours.github.io/daft-sources/DAFT-EN.pdf#page=98).

## Prepare the example

Run the cells in order in Jupyter or Google Colab. The first cell installs missing Python packages and downloads the figure sources and input data once. Subsequent runs use the cached download. Python 3.10 or later is required.

```python
from pathlib import Path
import hashlib
import importlib.util
import io
import os
import subprocess
import sys
import urllib.request
import zipfile

packages = {"numpy": "numpy", "scipy": "scipy", "matplotlib": "matplotlib",
            "PIL": "Pillow", "sympy": "sympy"}
missing = [package for module, package in packages.items()
           if importlib.util.find_spec(module) is None]
if missing:
    subprocess.check_call([sys.executable, "-m", "pip", "install", *missing])

archive_url = os.environ.get("DAFT_FIGURE_ARCHIVE", 'https://mathematical-tours.github.io/daft-sources/downloads/daft-figures-2e843bbe89f0.zip')
archive_sha = '2e843bbe89f0e274849b6896eb16bd9f7686521a8ca98f633240bfc06ea2d5b1'
workspace = Path(os.environ.get("DAFT_FIGURE_WORKDIR", "daft-figure-files")) / archive_sha[:12]
cache = workspace / "sources.zip"
workspace.mkdir(parents=True, exist_ok=True)
data = cache.read_bytes() if cache.exists() else urllib.request.urlopen(archive_url, timeout=60).read()
if hashlib.sha256(data).hexdigest() != archive_sha:
    raise ValueError("The figure-source download is incomplete or has changed. Download it again.")
cache.write_bytes(data)
with zipfile.ZipFile(io.BytesIO(data)) as archive:
    archive.extractall(workspace)
ROOT = (workspace / "daft-figures").resolve()
__file__ = str(ROOT / 'figures-code/chapter-fourier-applications/template-matching/generate.py')
print("Figure sources ready.")
```

## Generate the figure

This is the plotting code used by the book. Change its parameters and rerun this cell, then the display cell below, to explore the example.

```python
from pathlib import Path
import sys
ROOT=Path(__file__).resolve().parents[3]
sys.path.insert(0,str(ROOT/"tools"))
sys.path.insert(0,str(ROOT/"listings"))
from figure_style import *
from applications_data import flower
from image_processing import normalized_correlation
from scipy.signal import fftconvolve
from matplotlib.patches import Rectangle
f=flower(128);u,v,p=26,53,25;g=f[u:u+p,v:v+p]
raw=fftconvolve(f,g[::-1,::-1],mode='valid');normalized=normalized_correlation(f,g)
assert np.unravel_index(np.nanargmax(normalized),normalized.shape)==(u,v)
fig,axes=plt.subplots(2,2,figsize=(8,7.3))
axes[0,0].imshow(f,cmap='gray',vmin=0,vmax=1);axes[0,0].add_patch(Rectangle((v-.5,u-.5),p,p,fill=False,edgecolor=RED,lw=1.8));axes[0,0].set_title('Image and selected patch')
axes[0,1].imshow(g,cmap='gray',vmin=0,vmax=1);axes[0,1].set_title('Template')
im=axes[1,0].imshow(raw,cmap='magma');fig.colorbar(im,ax=axes[1,0],shrink=.78,label='raw correlation')
im=axes[1,1].imshow(normalized,cmap='RdBu_r',vmin=-1,vmax=1);fig.colorbar(im,ax=axes[1,1],shrink=.78,label='normalized correlation')
axes[1,0].set_title('Raw correlation');axes[1,1].set_title('Normalized correlation')
for ax in axes[0]:axes_box(ax);ax.set(xticks=[],yticks=[])
for ax in axes[1]:ax.plot(v,u,'x',color='#40ffbc',ms=8,mew=1.8);axes_box(ax,'patch column','patch row')
save(fig,'chapter-fourier-applications','template-matching')
```
