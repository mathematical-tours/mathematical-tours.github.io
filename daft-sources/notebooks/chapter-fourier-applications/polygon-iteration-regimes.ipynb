# Figure 4.6: Four regimes of polygon filtering

*The Discrete Algebra of the Fourier Transform*, Gabriel Peyré.

Four regimes of polygon filtering. Curves progress from the initial polygon to later iterates. Expanding active multipliers cause divergence; unit-modulus phases can keep the trajectory bounded without convergence.

[Read this figure in the book](https://mathematical-tours.github.io/daft-sources/DAFT-EN.pdf#page=88).

## Prepare the example

Run the cells in order in Jupyter or Google Colab. The first cell installs missing Python packages and downloads the figure sources and input data once. Subsequent runs use the cached download. Python 3.10 or later is required.

```python
from pathlib import Path
import hashlib
import importlib.util
import io
import os
import subprocess
import sys
import urllib.request
import zipfile

packages = {"numpy": "numpy", "scipy": "scipy", "matplotlib": "matplotlib",
            "PIL": "Pillow", "sympy": "sympy"}
missing = [package for module, package in packages.items()
           if importlib.util.find_spec(module) is None]
if missing:
    subprocess.check_call([sys.executable, "-m", "pip", "install", *missing])

archive_url = os.environ.get("DAFT_FIGURE_ARCHIVE", 'https://mathematical-tours.github.io/daft-sources/downloads/daft-figures-2e843bbe89f0.zip')
archive_sha = '2e843bbe89f0e274849b6896eb16bd9f7686521a8ca98f633240bfc06ea2d5b1'
workspace = Path(os.environ.get("DAFT_FIGURE_WORKDIR", "daft-figure-files")) / archive_sha[:12]
cache = workspace / "sources.zip"
workspace.mkdir(parents=True, exist_ok=True)
data = cache.read_bytes() if cache.exists() else urllib.request.urlopen(archive_url, timeout=60).read()
if hashlib.sha256(data).hexdigest() != archive_sha:
    raise ValueError("The figure-source download is incomplete or has changed. Download it again.")
cache.write_bytes(data)
with zipfile.ZipFile(io.BytesIO(data)) as archive:
    archive.extractall(workspace)
ROOT = (workspace / "daft-figures").resolve()
__file__ = str(ROOT / 'figures-code/chapter-fourier-applications/polygon-iteration-regimes/generate.py')
print("Figure sources ready.")
```

## Generate the figure

This is the plotting code used by the book. Change its parameters and rerun this cell, then the display cell below, to explore the example.

```python
from pathlib import Path
import sys
ROOT=Path(__file__).resolve().parents[3]
sys.path.insert(0,str(ROOT/"tools"))
sys.path.insert(0,str(ROOT/"listings"))
from figure_style import *
from applications_data import polygon,polygon_evolution,plot_polygon
z=polygon();n=len(z);steps=np.arange(26)
F=[np.full(n,.8,dtype=complex) for _ in range(4)]
F[0][:4]=[1.05*np.exp(1j*np.pi/15),1.1*np.exp(1j*np.pi/10),1.05*np.exp(1j*np.pi/8),1.15*np.exp(1j*np.pi/12)]
F[1][:4]=[1.2*np.exp(1j*np.pi/10),1.1*np.exp(1j*np.pi/8),1.15*np.exp(1j*np.pi/4),1.15*np.exp(1j*np.pi/6)]
F[2][:3]=[1,.9*np.exp(1j*np.pi/20),.9]
F[3][:3]=[np.exp(1j*np.pi/15),np.exp(1j*np.pi/8),np.exp(1j*np.pi/6)]
fig,axes=plt.subplots(2,2,figsize=(8.5,7))
for ax,multiplier,title in zip(axes.flat,F,['(a) Expanding modes','(b) Expansion and drift','(c) Convergence','(d) Bounded rotation']):
    trajectory=polygon_evolution(z,multiplier,steps)
    for i,poly in enumerate(trajectory):plot_polygon(ax,poly,color=plt.cm.viridis(.1+.75*i/25),lw=.65,alpha=.8)
    plot_polygon(ax,trajectory[-1],color=TEAL,lw=1.4)
    axes_box(ax,'real coordinate','imaginary coordinate');ax.set(title=title,aspect='equal')
axes[1,0].plot(z.mean().real,z.mean().imag,'o',color=RED,ms=4)
save(fig,'chapter-fourier-applications','polygon-iteration-regimes')
```
