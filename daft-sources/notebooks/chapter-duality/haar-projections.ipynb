# Figure 2.4: Successive Haar projections

*The Discrete Algebra of the Fourier Transform*, Gabriel Peyré.

Successive Haar projections onto $F_0$, $F_1$, and $F_2$ (left), and the orthonormal basis functions added at each stage (right).

[Read this figure in the book](https://mathematical-tours.github.io/daft-sources/DAFT-EN.pdf#page=39).

## Prepare the example

Run the cells in order in Jupyter or Google Colab. The first cell installs missing Python packages and downloads the figure sources and input data once. Subsequent runs use the cached download. Python 3.10 or later is required.

```python
from pathlib import Path
import hashlib
import importlib.util
import io
import os
import subprocess
import sys
import urllib.request
import zipfile

packages = {"numpy": "numpy", "scipy": "scipy", "matplotlib": "matplotlib",
            "PIL": "Pillow", "sympy": "sympy"}
missing = [package for module, package in packages.items()
           if importlib.util.find_spec(module) is None]
if missing:
    subprocess.check_call([sys.executable, "-m", "pip", "install", *missing])

archive_url = os.environ.get("DAFT_FIGURE_ARCHIVE", 'https://mathematical-tours.github.io/daft-sources/downloads/daft-figures-2e843bbe89f0.zip')
archive_sha = '2e843bbe89f0e274849b6896eb16bd9f7686521a8ca98f633240bfc06ea2d5b1'
workspace = Path(os.environ.get("DAFT_FIGURE_WORKDIR", "daft-figure-files")) / archive_sha[:12]
cache = workspace / "sources.zip"
workspace.mkdir(parents=True, exist_ok=True)
data = cache.read_bytes() if cache.exists() else urllib.request.urlopen(archive_url, timeout=60).read()
if hashlib.sha256(data).hexdigest() != archive_sha:
    raise ValueError("The figure-source download is incomplete or has changed. Download it again.")
cache.write_bytes(data)
with zipfile.ZipFile(io.BytesIO(data)) as archive:
    archive.extractall(workspace)
ROOT = (workspace / "daft-figures").resolve()
__file__ = str(ROOT / 'figures-code/chapter-duality/haar-projections/generate.py')
print("Figure sources ready.")
```

## Generate the figure

This is the plotting code used by the book. Change its parameters and rerun this cell, then the display cell below, to explore the example.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path(__file__).resolve().parents[3] / "tools"))
sys.path.insert(0, str(Path(__file__).resolve().parents[3] / "listings"))
from figure_style import *
x=(np.arange(2048)+.5)/2048
f=.25+.8*x+.12*np.sin(2*np.pi*x)+.14*np.sin(5*np.pi*x)
fig,axs=plt.subplots(3,2,figsize=(7.1,5.1),sharex=True)
for row,J in enumerate([0,1,2]):
    ax=axs[row,0];size=len(x)//2**J
    p=np.repeat(f.reshape(2**J,size).mean(axis=1),size)
    ax.plot(x,f,color=MUTED,lw=1,label='$f$')
    ax.step(np.r_[0,x,1],np.r_[p[0],p,p[-1]],where='mid',color=TEAL,label=rf'$P_{{F_{J}}}f$')
    ax.set(ylim=(.1,1.25),yticks=[.2,.6,1.0])
    ax.legend(loc='upper left',frameon=False,fontsize=9,ncol=2)
    ax=axs[row,1]
    if J==0:
        ax.plot([0,1],[1,1],color=TEAL,label=r'$\psi_0$')
        ax.set(ylim=(-.15,1.45),yticks=[0,1])
    else:
        j=J-1
        for k,color in zip(range(2**j),[TEAL,RED]):
            left=k/2**j;mid=(k+.5)/2**j;right=(k+1)/2**j
            xx=[0,left,left,mid,mid,right,right,1]
            y=[0,0,2**(j/2),2**(j/2),-2**(j/2),-2**(j/2),0,0]
            ax.plot(xx,y,color=color,label=rf'$\psi_{{{j},{k}}}$')
        ax.set(ylim=(-1.8,2.1),yticks=[-1,0,1])
    ax.legend(loc='upper right',frameon=False,fontsize=9,ncol=2)
    for col in range(2):
        axs[row,col].set(xlim=(0,1),xticks=[0,.5,1]);axes_box(axs[row,col],'$x$' if row==2 else None)
axs[0,0].set_title('Orthogonal projection');axs[0,1].set_title('New basis functions')
save(fig,'chapter-duality','haar-projections')
```
