# Figure 5.3: The notch-filter transfer function

*The Discrete Algebra of the Fourier Transform*, Gabriel Peyré.

Magnitude of the rational notch-filter transfer function in the complex plane. Values above two are clipped to reveal the zeros and pole neighborhoods. The causal Laurent series converges for $|z|>0.9$.

[Read this figure in the book](https://mathematical-tours.github.io/daft-sources/DAFT-EN.pdf#page=118).

## Prepare the example

Run the cells in order in Jupyter or Google Colab. The first cell installs missing Python packages and downloads the figure sources and input data once. Subsequent runs use the cached download. Python 3.10 or later is required.

```python
from pathlib import Path
import hashlib
import importlib.util
import io
import os
import subprocess
import sys
import urllib.request
import zipfile

packages = {"numpy": "numpy", "scipy": "scipy", "matplotlib": "matplotlib",
            "PIL": "Pillow", "sympy": "sympy"}
missing = [package for module, package in packages.items()
           if importlib.util.find_spec(module) is None]
if missing:
    subprocess.check_call([sys.executable, "-m", "pip", "install", *missing])

archive_url = os.environ.get("DAFT_FIGURE_ARCHIVE", 'https://mathematical-tours.github.io/daft-sources/downloads/daft-figures-2e843bbe89f0.zip')
archive_sha = '2e843bbe89f0e274849b6896eb16bd9f7686521a8ca98f633240bfc06ea2d5b1'
workspace = Path(os.environ.get("DAFT_FIGURE_WORKDIR", "daft-figure-files")) / archive_sha[:12]
cache = workspace / "sources.zip"
workspace.mkdir(parents=True, exist_ok=True)
data = cache.read_bytes() if cache.exists() else urllib.request.urlopen(archive_url, timeout=60).read()
if hashlib.sha256(data).hexdigest() != archive_sha:
    raise ValueError("The figure-source download is incomplete or has changed. Download it again.")
cache.write_bytes(data)
with zipfile.ZipFile(io.BytesIO(data)) as archive:
    archive.extractall(workspace)
ROOT = (workspace / "daft-figures").resolve()
__file__ = str(ROOT / 'figures-code/chapter-related-transforms/notch-transfer-surface/generate.py')
print("Figure sources ready.")
```

## Generate the figure

This is the plotting code used by the book. Change its parameters and rerun this cell, then the display cell below, to explore the example.

```python
from pathlib import Path
import sys
ROOT=Path(__file__).resolve().parents[3]
sys.path.insert(0,str(ROOT/"tools"))
sys.path.insert(0,str(ROOT/"listings"))
from figure_style import *
from related_transform_data import notch
x=np.linspace(-1.45,1.45,121);X,Y=np.meshgrid(x,x);Z=abs(notch(X+1j*Y));limit=2
fig=plt.figure(figsize=(6.5,4.9));ax=fig.add_subplot(projection='3d')
surf=ax.plot_surface(X,Y,np.minimum(Z,limit),cmap='viridis',vmin=0,vmax=limit,rcount=90,ccount=90,linewidth=.16,edgecolor='#344b5550',antialiased=True)
t=np.linspace(0,2*np.pi,400);ax.plot(np.cos(t),np.sin(t),np.zeros_like(t),color=TEAL,lw=1.3)
ax.set(xlabel=r'$\operatorname{Re}z$',ylabel=r'$\operatorname{Im}z$',zlabel='$|H(z)|$',zlim=(0,limit),xticks=[-1,0,1],yticks=[-1,0,1]);ax.view_init(26,-124)
ax.set_box_aspect((1,1,.72));fig.colorbar(surf,ax=ax,shrink=.56,pad=.04,label='magnitude (clipped)')
save(fig,'chapter-related-transforms','notch-transfer-surface')
```
